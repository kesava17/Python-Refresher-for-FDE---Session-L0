# NumPy: detailed scenarios

A runnable tour of NumPy for analytics. Each section is a scenario you can run and tweak. Self-contained, no data files needed. See **numpy_cheatsheet.md** for the full reference.

```python
import numpy as np
rng = np.random.default_rng(0)
```

## Creating arrays

```python
print("arange:", np.arange(0, 10, 2))
print("linspace:", np.linspace(0, 1, 5))
print("zeros:")
print(np.zeros((2, 3)))
print("eye:")
print(np.eye(3))
print("random:")
print(rng.random((2, 3)).round(3))
```

## Inspecting an array

```python
m = rng.normal(size=(4, 5))
print("shape", m.shape, "ndim", m.ndim, "size", m.size, "dtype", m.dtype)
```

## Indexing and slicing: ranges, boolean masks, fancy indexing

```python
print("row 0:", m[0].round(2))
print("col 1:", m[:, 1].round(2))
print("positives, first 5:", m[m > 0][:5].round(2))
print("where > 0 -> 1 else 0, first 2 rows:")
print(np.where(m > 0, 1, 0)[:2])
```

## Reshaping and combining

```python
print("reshape:", m.reshape(5, 4).shape)
print("ravel:", m.ravel().shape)
print("transpose:", m.T.shape)
print("vstack:", np.vstack([m, m]).shape)
print("concat axis=1:", np.concatenate([m, m], axis=1).shape)
```

## Element-wise math (ufuncs)

```python
print("sqrt of abs, row 0:", np.sqrt(np.abs(m[0])).round(2))
print("clip to [-1, 1], row 0:", np.clip(m[0], -1, 1).round(2))
```

## Aggregations and the axis argument

axis=0 reduces down the rows (per column); axis=1 reduces across columns (per row).

```python
print("overall mean:", m.mean().round(3))
print("per-column mean:", m.mean(axis=0).round(2))
print("per-row std:", m.std(axis=1).round(2))
print("90th percentile:", np.percentile(m, 90).round(2))
```

## Broadcasting: standardise each column

```python
z = (m - m.mean(axis=0)) / m.std(axis=0)
print("column means near 0:", z.mean(axis=0).round(3))
print("column stds near 1:", z.std(axis=0).round(3))
```

## Linear algebra

```python
A = rng.normal(size=(3, 3))
b = rng.normal(size=3)
x = np.linalg.solve(A, b)
print("solve Ax=b, max residual:", np.abs(A @ x - b).max().round(6))
print("eigenvalues:", np.round(np.linalg.eigvals(A), 2))
```

## Analytics scenario: a price series

Simulate a random-walk price, then compute returns, annualised volatility, a moving average, and the drawdown.

```python
steps = rng.normal(0.0005, 0.01, size=252)
prices = 100 * np.cumprod(1 + steps)
returns = prices[1:] / prices[:-1] - 1
print("annualised vol:", round(returns.std() * np.sqrt(252), 3))

from numpy.lib.stride_tricks import sliding_window_view
sma20 = sliding_window_view(prices, 20).mean(axis=1)
print("first SMA(20):", round(float(sma20[0]), 2))

cum = np.cumprod(1 + returns)
drawdown = cum / np.maximum.accumulate(cum) - 1
print("max drawdown:", round(float(drawdown.min()), 3))
```
